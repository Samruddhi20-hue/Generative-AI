```python
"""
Prompt Engineering Approaches Implementation

Aim:
To implement and analyze different prompt engineering approaches:
- Interview Prompting
- Chain-of-Thought (CoT)
- Tree-of-Thought (ToT)

Also, to implement Zero-Shot and Few-Shot prompting and compare their results.

This implementation is done using simulated LLM responses for academic purposes.
"""
```

```python
print("INTERVIEW PROMPTING APPROACH\n")

questions = [
    "What is the cost of one pen?",
    "What is the cost of one notebook?",
    "How many pens are bought?",
    "How many notebooks are bought?",
    "Calculate the total cost."
]

answers = [
    "The cost of one pen is ₹10.",
    "The cost of one notebook is ₹40.",
    "3 pens are bought.",
    "2 notebooks are bought.",
    "Total cost = (3 × 10) + (2 × 40) = ₹110"
]

for q, a in zip(questions, answers):
    print("Q:", q)
    print("A:", a)
    print()
```

### Interview Prompting

In this approach, the model is guided step-by-step through structured questions,
similar to an interview process. It improves clarity and user understanding but may
be slower for complex reasoning tasks.


```python
print("CHAIN-OF-THOUGHT (CoT) PROMPTING\n")

cot_output = """
Step 1: Cost of pens = 3 × 10 = ₹30
Step 2: Cost of notebooks = 2 × 40 = ₹80
Step 3: Total cost = 30 + 80 = ₹110
"""

print(cot_output)
```

## Problem Statement

To demonstrate how different prompt engineering approaches influence reasoning and output
quality of Large Language Models using a simple arithmetic and language translation task.


### Chain-of-Thought Prompting

Chain-of-Thought prompting encourages the model to explicitly show intermediate
reasoning steps before producing the final answer. This improves accuracy in
logical and mathematical problem-solving tasks.


```python
print("TREE-OF-THOUGHT (ToT) PROMPTING\n")

reasoning_paths = {
    "Path 1 (Direct Calculation)": "Total = ₹110",
    "Path 2 (Item-wise Calculation)": "(3×10) + (2×40) = ₹110",
    "Path 3 (Unit Method)": "Pen total ₹30 + Notebook total ₹80 = ₹110"
}

for path, result in reasoning_paths.items():
    print(path, "->", result)

print("\nFinal Selected Answer: ₹110")
```

### Tree-of-Thought Prompting

Tree-of-Thought prompting explores multiple reasoning paths before selecting the
best solution. This approach is useful in complex decision-making and planning tasks.


```python
print("ZERO-SHOT PROMPTING\n")

prompt = "Translate to French: I love machine learning"
output = "J'aime l'apprentissage automatique."

print("Prompt:", prompt)
print("Output:", output)
```

```python
print("FEW-SHOT PROMPTING\n")

prompt = """
English: Hello
French: Bonjour

English: Thank you
French: Merci

English: I love machine learning
French:
"""

output = "J'aime l'apprentissage automatique."

print("Prompt:", prompt)
print("Output:", output)
```

## Comparison and Analysis

### Prompt Engineering Approaches

| Approach | Reasoning Style | Accuracy | Complexity | Applications |
|--------|---------------|----------|------------|--------------|
| Interview | Guided Q&A | Medium | Low | Education, Chatbots |
| CoT | Step-by-step | High | Medium | Math, Logic |
| ToT | Multi-path | Very High | High | AI Planning |

### Zero-Shot vs Few-Shot Prompting

| Feature | Zero-Shot | Few-Shot |
|-------|----------|----------|
| Examples | None | Few |
| Accuracy | Medium | High |
| Control | Low | High |


## Conclusion

Prompt Engineering significantly improves the performance of Large Language Models.
Chain-of-Thought and Tree-of-Thought prompting enhance reasoning accuracy, while
Few-Shot prompting provides more consistent results compared to Zero-Shot prompting.
